```python
import numpy as np
from matplotlib import pyplot as plt
from pint import UnitRegistry; AssignQuantity = UnitRegistry().Quantity
```

```python
R = AssignQuantity(8.314,'J/mol/K')
Trange = np.linspace(265,273,20)
Trange = AssignQuantity(Trange,'K')
h = AssignQuantity(6.626e-34,'J*s')
kB = AssignQuantity(1.38e-23,'J/K')
```

```python
DeltaG = AssignQuantity(114,'kJ/mol')
time = h/(kB*Trange)*np.exp(DeltaG/(R*Trange))
time.ito('year')
print(time)

plt.figure()
plt.plot(Trange.magnitude,time.magnitude)
plt.grid(True)
plt.xlabel('Temperature ('+str(Trange.units)+')')
plt.ylabel('time ('+str(time.units)+')')
plt.title(r'When $\Delta G = $'+str(DeltaG))
```

```python
distance = np.array([3.6,3.8,4,4.2,4.4]); distance = AssignQuantity(distance,'angstrom')
Energy_of_activation = np.array([134,132,131,114,127]); Energy_of_activation = AssignQuantity(Energy_of_activation,'kJ/mol')
plt.figure()
plt.plot(distance.magnitude,Energy_of_activation.magnitude)
plt.grid(True)
plt.xlabel('O7-O9 distance ('+str(distance.units)+')')
plt.ylabel(r'$E_a$ ('+str(Energy_of_activation.units)+')')
# plt.title(r'When $\Delta G = $'+str(DeltaG))
```
