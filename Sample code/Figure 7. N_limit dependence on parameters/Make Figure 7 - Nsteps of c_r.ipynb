### Exploring the dependence of $N_{steps}^{limit}$ on $c_r$

```python
# Get default parameters
GIfile = '../GI parameters - Table 1.nml'
sigmaI_corner, sigma0, c_r_ref, sigmaI_QLC_ref, Nbar, Nstar, h_layer, D, nu_kin_ref, L, nx_crystal, t_eq_ref, x_QLC, DoverL2pi2, Doverdeltax2 = \
    QLC.get_system_parameters(GIfile, AssignQuantity)
odemethod, tkeep_1Darr_ref, Ntot_init_1D, NQLL_init_1D = \
    QLC.get_runtime_parameters(GIfile, nx_crystal, Nstar, Nbar, AssignQuantity)
```

### Looping over c_r

```python
# Adjust timeline as necessary
tkeep_1Darr = tkeep_1Darr_ref*7/3

# Set up the loop
nruns = 10
c_r_range = np.linspace(c_r_ref*0.25,c_r_ref*1.1,nruns)
print('Going to loop over c_r_range =', c_r_range)

# Keeping other values constant for this looop
t_eq = t_eq_ref
nu_kin = nu_kin_ref

# A place to store results for later
Nsteps_limit_cr = np.zeros(np.shape(c_r_range))

# Loop over the variations
for i_cr, c_r in enumerate(c_r_range):

    sigmaI_QLC = sigmaI_corner*(c_r*(x_QLC/L)**2+1-c_r)
    Ntotkeep_1D_FT, NQLLkeep_1D_FT = QLC.run_f1d_FT(\
               NQLL_init_1D,Ntot_init_1D,tkeep_1Darr,\
               Nbar, Nstar, sigma0, nu_kin, DoverL2pi2, t_eq, sigmaI_QLC,\
               AssignQuantity,\
               verbose=0, odemethod='RK45') # RK45 allows integration of complex functions (but LSODA does not)
    Nicekeep_1D_FT = Ntotkeep_1D_FT - NQLLkeep_1D_FT
    
    # Graphical output
    fig = QLC.report_1d_growth_results_with_sigma(\
        x_QLC,tkeep_1Darr.to('millisecond'),NQLLkeep_1D_FT,Ntotkeep_1D_FT,Nicekeep_1D_FT,sigmaI_QLC.to('percent'))

    Nsteps_limit_cr[i_cr] = np.max(Ntotkeep_1D_FT[-1,:])-np.min(Ntotkeep_1D_FT[-1,:])

plt.figure()
plt.plot(c_r_range,Nsteps_limit_cr)
```

```python
# Saving
data = [c_r_range, Nsteps_limit_cr]
crfile = 'Nsteps_limit(c_r).txt'
np.savetxt(crfile, data, header="c_r, Nsteps_limit", comments="#")


# c_r_loaded, Nsteps_loaded = np.loadtxt(crfile)
# print('c_r_loaded',c_r_loaded)
# print('Nsteps_loaded',Nsteps_loaded)
```
